```python
# tanpa solid
# [ISP VIOLATION]
# Interface terlalu besar


class PegawaiInterface:

    def bekerja(self):
        raise NotImplementedError

    def coding(self):
        raise NotImplementedError

    def desain(self):
        raise NotImplementedError

    def presentasi(self):
        raise NotImplementedError

    def mengelola_server(self):
        raise NotImplementedError



# [SRP VIOLATION]
# Class Pegawai mengurus terlalu banyak hal:
# - data pegawai
# - database
# - email
# - cetak slip
class Pegawai(PegawaiInterface):

    def __init__(self, id_pegawai, nama, jabatan, gaji_pokok):
        self.id_pegawai = id_pegawai
        self.nama = nama
        self.jabatan = jabatan
        self.gaji_pokok = gaji_pokok

    def simpan_ke_database(self):
        print(f"[DATABASE] Menyimpan data {self.nama} ke database...")

    def kirim_email(self):
        print(f"[EMAIL] Mengirim email ke {self.nama}...")

    def cetak_slip_gaji(self, total_gaji):
        print("================================")
        print("SLIP GAJI PEGAWAI")
        print("================================")
        print(f"Nama   : {self.nama}")
        print(f"Jabatan: {self.jabatan}")
        print(f"Gaji   : {total_gaji}")
        print("================================")

    def bekerja(self):
        print(f"{self.nama} sedang bekerja.")

    def coding(self):
        print(f"{self.nama} sedang coding.")

    def desain(self):
        print(f"{self.nama} sedang mendesain.")

    def presentasi(self):
        print(f"{self.nama} sedang presentasi.")

    def mengelola_server(self):
        print(f"{self.nama} sedang mengelola server.")



# [OCP VIOLATION]
# Setiap jabatan baru -> harus ubah class ini
class KalkulatorGaji:

    def hitung_gaji(self, pegawai):

        if pegawai.jabatan == "manager":
            bonus = 5000000

        elif pegawai.jabatan == "programmer":
            bonus = 3000000

        elif pegawai.jabatan == "designer":
            bonus = 2500000

        elif pegawai.jabatan == "intern":
            bonus = 500000

        else:
            bonus = 0

        return pegawai.gaji_pokok + bonus



# [LSP VIOLATION]
# HRD tidak bisa coding, tapi parent class mengharuskan
class HRD(Pegawai):

    def coding(self):
        raise Exception("HRD tidak bisa coding!")



# [DIP VIOLATION]
# Sistem bergantung langsung pada DatabaseMySQL
class DatabaseMySQL:

    def connect(self):
        print("Terhubung ke database MySQL")


class SistemPerusahaan:

    def __init__(self):

        # Hardcoded dependency
        self.database = DatabaseMySQL()

    def jalankan(self):
        self.database.connect()
        print("Sistem perusahaan berjalan")



# PROGRAM UTAMA


pegawai1 = Pegawai(
    id_pegawai="PG001",
    nama="Andi",
    jabatan="programmer",
    gaji_pokok=7000000
)

pegawai2 = HRD(
    id_pegawai="PG002",
    nama="Siti",
    jabatan="manager",
    gaji_pokok=9000000
)


# OCP VIOLATION
kalkulator = KalkulatorGaji()

gaji1 = kalkulator.hitung_gaji(pegawai1)
gaji2 = kalkulator.hitung_gaji(pegawai2)

pegawai1.cetak_slip_gaji(gaji1)
pegawai2.cetak_slip_gaji(gaji2)

print()


# LSP VIOLATION
try:
    pegawai2.coding()

except Exception as e:
    print("ERROR:", e)

print()


# DIP VIOLATION
sistem = SistemPerusahaan()
sistem.jalankan()
```

```python
# Dengan solid

from abc import ABC, abstractmethod

# [ISP]
# Interface dipisah menjadi kecil dan spesifik
class BisaBekerja(ABC):

    @abstractmethod
    def bekerja(self):
        pass


class BisaCoding(ABC):

    @abstractmethod
    def coding(self):
        pass


class BisaDesain(ABC):

    @abstractmethod
    def desain(self):
        pass


class BisaPresentasi(ABC):

    @abstractmethod
    def presentasi(self):
        pass


class BisaMengelolaServer(ABC):

    @abstractmethod
    def mengelola_server(self):
        pass


# [SRP]
# Class Pegawai hanya menyimpan data pegawai
class Pegawai:

    def __init__(self, id_pegawai, nama, gaji_pokok):
        self.id_pegawai = id_pegawai
        self.nama = nama
        self.gaji_pokok = gaji_pokok


# [LSP]
# Setiap subclass hanya implement kemampuan yang sesuai
class Programmer(Pegawai, BisaBekerja, BisaCoding):

    def bekerja(self):
        print(f"{self.nama} sedang bekerja sebagai programmer.")

    def coding(self):
        print(f"{self.nama} sedang coding aplikasi.")


class Designer(Pegawai, BisaBekerja, BisaDesain):

    def bekerja(self):
        print(f"{self.nama} sedang bekerja sebagai designer.")

    def desain(self):
        print(f"{self.nama} sedang membuat desain UI.")


class Manager(Pegawai, BisaBekerja, BisaPresentasi):

    def bekerja(self):
        print(f"{self.nama} sedang bekerja sebagai manager.")

    def presentasi(self):
        print(f"{self.nama} sedang presentasi proyek.")


class HRD(Pegawai, BisaBekerja):

    def bekerja(self):
        print(f"{self.nama} sedang mengurus administrasi karyawan.")

# [OCP]
# Strategy Pattern
# Tambah jabatan baru cukup tambah class baru
class StrategiGaji(ABC):

    @abstractmethod
    def hitung_gaji(self, pegawai):
        pass


class GajiProgrammer(StrategiGaji):

    def hitung_gaji(self, pegawai):
        bonus = 3000000
        return pegawai.gaji_pokok + bonus


class GajiDesigner(StrategiGaji):

    def hitung_gaji(self, pegawai):
        bonus = 2500000
        return pegawai.gaji_pokok + bonus


class GajiManager(StrategiGaji):

    def hitung_gaji(self, pegawai):
        bonus = 5000000
        return pegawai.gaji_pokok + bonus


class GajiHRD(StrategiGaji):

    def hitung_gaji(self, pegawai):
        bonus = 2000000
        return pegawai.gaji_pokok + bonus



# Calculator khusus hitung gaji
class KalkulatorGaji:

    def __init__(self, strategi: StrategiGaji):
        self.strategi = strategi

    def hitung(self, pegawai):
        return self.strategi.hitung_gaji(pegawai)


# [SRP]
# Class khusus database
class PenyimpananPegawai:

    def simpan(self, pegawai):
        print(f"[DATABASE] Data {pegawai.nama} berhasil disimpan.")



# [SRP]
# Class khusus email
class PengirimEmail:

    def kirim(self, pegawai):
        print(f"[EMAIL] Email berhasil dikirim ke {pegawai.nama}.")



# [SRP]
# Class khusus cetak slip gaji
class PencetakSlipGaji:

    def cetak(self, pegawai, total_gaji):

        print("================================")
        print("SLIP GAJI PEGAWAI")
        print("================================")
        print(f"ID Pegawai : {pegawai.id_pegawai}")
        print(f"Nama       : {pegawai.nama}")
        print(f"Gaji Total : Rp {total_gaji}")
        print("================================")



# [DIP]
# Abstraksi database
class Database(ABC):

    @abstractmethod
    def connect(self):
        pass


class DatabaseMySQL(Database):

    def connect(self):
        print("Terhubung ke database MySQL")


class DatabasePostgreSQL(Database):

    def connect(self):
        print("Terhubung ke database PostgreSQL")



# [DIP]
# Depend pada abstraksi, bukan class konkret
class SistemPerusahaan:

    def __init__(self, database: Database):
        self.database = database

    def jalankan(self):
        self.database.connect()
        print("Sistem perusahaan berjalan")



# PROGRAM UTAMA


pegawai1 = Programmer(
    id_pegawai="PG001",
    nama="Andi",
    gaji_pokok=7000000
)

pegawai2 = HRD(
    id_pegawai="PG002",
    nama="Siti",
    gaji_pokok=9000000
)

pegawai3 = Manager(
    id_pegawai="PG003",
    nama="Budi",
    gaji_pokok=10000000
)


# Menjalankan pekerjaan


pegawai1.bekerja()
pegawai1.coding()

print()

pegawai2.bekerja()

print()

pegawai3.bekerja()
pegawai3.presentasi()

print()


# Menghitung gaji


kalkulator_programmer = KalkulatorGaji(GajiProgrammer())
kalkulator_hrd = KalkulatorGaji(GajiHRD())
kalkulator_manager = KalkulatorGaji(GajiManager())

gaji1 = kalkulator_programmer.hitung(pegawai1)
gaji2 = kalkulator_hrd.hitung(pegawai2)
gaji3 = kalkulator_manager.hitung(pegawai3)


# Simpan database


penyimpanan = PenyimpananPegawai()

penyimpanan.simpan(pegawai1)
penyimpanan.simpan(pegawai2)
penyimpanan.simpan(pegawai3)

print()


# Kirim email


email = PengirimEmail()

email.kirim(pegawai1)
email.kirim(pegawai2)
email.kirim(pegawai3)

print()


# Cetak slip gaji


printer = PencetakSlipGaji()

printer.cetak(pegawai1, gaji1)
printer.cetak(pegawai2, gaji2)
printer.cetak(pegawai3, gaji3)

print()


# Dependency Injection


sistem_mysql = SistemPerusahaan(DatabaseMySQL())
sistem_postgres = SistemPerusahaan(DatabasePostgreSQL())

sistem_mysql.jalankan()
sistem_postgres.jalankan()
```
